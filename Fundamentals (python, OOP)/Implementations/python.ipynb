```python
# Positional and keyword arguments
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Amr"))                  # positional + default
print(greet("Amr", "Welcome"))       # positional
print(greet(name="Amr", greeting="Hi"))  # keyword

```

```python
old = 69

# *args and **kwargs
def summarize(*numbers, **options):
    total = sum(numbers)
    title = options.get("label", "XD")
    return f"{title}: {total}"

print(summarize(1, 2, 3))
print(summarize(1, 2, 3, label="Sum"))
print(summarize(1, 2, 3, Total="nibba"))
```

```python


# Scope: LEGB (Local, Enclosing, Global, Built-in)
scope_value = "global"

def outer():
    scope_value = "enclosing"

    def inner():
        local_value = "local"
        print(local_value)   # Local
        print(scope_value)   # Enclosing
        print(old)           # Global
        print(len("Python")) # Built-in

    inner()
outer()
print(scope_value)  # Global
```

```python
# Mutable defaults
def append_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(append_item("a"))
print(append_item("b"))
```

```python

# First-class functions, closures, lambda, higher-order functions, decorators
def apply(function, value):
    return function(value)

print(apply(lambda n: n * 2, 5))

def multiplier(factor):
    def multiply(value):
        return value * factor
    return multiply

double = multiplier(2)
print(double(6))

def log_call(function):
    @functools.wraps(function)
    def wrapper(*args, **kwargs):
        print(f"Calling {function.__name__}")
        return function(*args, **kwargs)
    return wrapper

@log_call
def add(a, b):
    return a + b

print(add(2, 3))

# Iterables, iterators, generators, lazy evaluation
numbers = [1, 2, 3]
iterator = iter(numbers)
print(next(iterator), next(iterator))

def countdown(n):
    while n:
        yield n
        n -= 1

print(list(countdown(3)))
lazy_squares = (n * n for n in range(4))
print(list(lazy_squares))

# Context managers and cleanup
class ManagedResource:
    def __enter__(self):
        print("resource opened")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("resource closed")
        return False

with ManagedResource():
    print("using resource")

# Exceptions and exception chaining
class ValidationError(ValueError):
    pass

try:
    value = int("not a number")
except ValueError as error:
    raise ValidationError("Input must be an integer") from error
```
